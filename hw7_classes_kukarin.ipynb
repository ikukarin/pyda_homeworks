```python
import requests

def get_name_by_max_value():
    max_value = 0
    name = ''
    r = requests.get('https://www.cbr-xml-daily.ru/daily_json.js')

    for currency in r.json()['Valute'].values():
        if currency['Value'] > max_value:
            max_value = currency['Value']
            name = currency['Name']
    return name


print(get_name_by_max_value())
```
