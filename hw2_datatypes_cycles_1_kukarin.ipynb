```python
word = input("Введите слово: ")
length = len(word)

if length % 2 == 0:
    print(word[length // 2 - 1] + word[length // 2])
else:
    print(word[length // 2])
```

```python
boys = ['Василий', 'Федор', 'Эдуард', 'Вениамин', 'Петр']
girls = ['Василиса', 'Февронья', 'Юлия', 'Анастасия', 'Надежда']

sorted_boys = sorted(boys)
sorted_girls = sorted(girls)

if len(sorted_boys) == len(sorted_girls):
    print("Идеальные пары: ")
    for boy, girl in zip(sorted_boys, sorted_girls):
        print(f"{boy} и {girl}")
else:
    print("Внимание, кто-то может остаться без пары!")
```

```python
import re

numbers = ['АБ03РЕ123', 'Х895МН98', 'М131НС1234']

for item in numbers:
    result = re.match('[АВЕКМНОРСТУХ]{1}\d{3}[АВЕКМНОРСТУХ]{2}\d{2,3}$', item)
    if result:
        print(f"Номер {item[:6]} валиден. Регион: {item[6:]}")
    else:
        print(f"Номер {item} не валиден")
```

```python
import re

stream = [
    'user4,2021-01-01;3',
    'user3,2022-01-07;4',
    'user2,2022-03-29;1',
    'user1,2020-04-04;13',
    'user2,2022-01-05;7',
    'user1,2021-06-14;4',
    'user3,2022-07-02;10',
    'user4,2021-03-21;19',
    'user4,2022-03-22;4',
    'user4,2022-04-22;8',
    'user4,2021-05-03;9',
    'user4,2022-05-11;11'
]
views = 0
users = []
for item in stream:
    user = re.search('user\d*', item).group()
    view = re.search('(\d*).$', item).group()
    if user not in users:
        users.append(user)
    views += int(view)
average_views = views / len(users)
print(f"Среднее количество просмотров на уникального пользователя: {round(average_views, 2)}")
```
