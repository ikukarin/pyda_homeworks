```python
queries = [
    'смотреть сериалы онлайн',
    'новости спорта',
    'афиша кино',
    'курс',
    'что будет, если поставить знаки препинания?',
    'сериалы этим летом',
    'курс по питону',
    'сериалы про спорт',
    'Python для чайников скачать бесплатно без регистрации и смс'
]

res = {}

for query in queries:
    count = len(query.split())
    res.setdefault(count, 0)
    res[count] += 1

print(res)     # получили словарь, ключи которого - длина запроса, а значения - количество таких запросов

for length, count in res.items():
    print(f'Поисковых запросов, содержащих {length} слов(а): {round((count*100)/len(queries), 2)}%')
```

```python
results = {
    'vk': {'revenue': 103, 'cost': 98},
    'yandex': {'revenue': 179, 'cost': 153},
    'facebook': {'revenue': 103, 'cost': 110},
    'adwords': {'revenue': 35, 'cost': 34},
    'twitter': {'revenue': 11, 'cost': 24},
}

for value in results.values():
    value['ROI'] = round((((value['revenue']/value['cost'])-1)*100), 2)
print(results)
```

```python
stats = {'facebook': 55,
         'yandex': 115,
         'vk': 120,
         'google': 100,
         'email': 42,
         'ok': 98}

company = [key for key in stats if stats[key] == max(stats.values())]
print(f'Максимальный объем продаж на рекламном канале: {"".join(company)}')
```
