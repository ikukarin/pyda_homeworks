```python
import re

documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}


def get_owner_name(number):
    for doc in documents:
        if doc['number'] == number:
            return f"Владелец документа: {doc['name']}"
    return 'Документ не найден в базе'


def get_directory(number):
    for dir_, nums in directories.items():
        if number in nums:
            return f'Документ хранится на полке: {dir_}'
    return 'Документ не найден в базе'


def list_all_documents():
    for document in documents:
        dir_ = re.search('\d+', get_directory(document['number'])).group()
        print(f"№: {document['number']}, тип: {document['type']}, владелец: {document['name']}, полка хранения: {dir_}")


def is_dir_exist(dir_number):
    """Проверяет, есть ли полка."""
    return dir_number in directories


def is_dir_empty(dir_number):
    """Проверяет, пуста ли полка."""
    return len(directories[dir_number]) == 0


def add_new_dir(dir_number):
    """Добавляет новую полку."""
    if not is_dir_exist(dir_number):
        directories[dir_number] = []
        print(f"Полка добавлена. Текущий перечень полок: {', '.join(directories.keys())}.")
    else:
        print(f"Такая полка уже существует. Текущий перечень полок: {', '.join(directories.keys())}.")


def remove_dir(dir_number):
    """Удаляет полку, если она пуста."""
    if is_dir_exist(dir_number):
        if is_dir_empty(dir_number):
            del directories[dir_number]
            print(f"Полка удалена. Текущий перечень полок: {', '.join(directories.keys())}.")
        else:
            print(f"На полке есть документы, удалите их перед удалением полки. Текущий перечень полок: {', '.join(directories.keys())}.")
    else:
        print(f"Такой полки не существует. Текущий перечень полок: {', '.join(directories.keys())}.")


def main(command):
    while command != 'q':
        command = input('Введите команду: ')
        if command == 'p':
            number = input('Введите номер документа: ')
            print(get_owner_name(number))
        elif command == 's':
            number = input('Введите номер документа: ')
            print(get_directory(number))
        elif command == 'l':
            list_all_documents()
        elif command == 'ads':
            dir_number = input('Введите номер полки: ')
            add_new_dir(dir_number)
        elif command == 'ds':
            dir_number = input('Введите номер полки: ')
            remove_dir(dir_number)


ret_code = main('')
exit(ret_code)
```
