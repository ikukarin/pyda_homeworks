```python
day = int(input("Введите день: "))
month = str(input("Введите месяц: "))

match month:
    case "Январь" | "январь" | "01" | "1":
        if 1 <= day <= 20: print("Козерог")
        elif 21 <= day <= 31: print("Водолей")
        else: print("Нет такой даты")

    case "Февраль" | "февраль" | "02" | "2":
        if 1 <= day <= 20: print("Водолей")
        elif 21 <= day <= 29: print("Рыбы")
        else: print("Нет такой даты")

    case "Март" | "март" | "03" | "3":
        if 1 <= day <= 20: print("Рыбы")
        elif 21 <= day <= 31: print("Овен")
        else: print("Нет такой даты")

    case "Апрель" | "апрель" | "04" | "4":
        if 1 <= day <= 20: print("Овен")
        elif 21 <= day <= 30: print("Телец")
        else: print("Нет такой даты")

    case "Май" | "май" | "05" | "5":
        if 1 <= day <= 20: print("Телец")
        elif 21 <= day <= 31: print("Близнецы")
        else: print("Нет такой даты")

    case "Июнь" | "июнь" | "06" | "6":
        if 1 <= day <= 21: print("Близнецы")
        elif 22 <= day <= 30: print("Рак")
        else: print("Нет такой даты")

    case "Июль" | "июль" | "07" | "7":
        if 1 <= day <= 22: print("Рак")
        elif 23 <= day <= 31: print("Лев")
        else: print("Нет такой даты")

    case "Август" | "август" | "08" | "8":
        if 1 <= day <= 23: print("Лев")
        elif 24 <= day <= 31: print("Дева")
        else: print("Нет такой даты")

    case "Сентябрь" | "сентябрь" | "09" | "9":
        if 1 <= day <= 23: print("Дева")
        elif 24 <= day <= 30: print("Весы")
        else: print("Нет такой даты")

    case "Октябрь" | "октябрь" | "10" | "10":
        if 1 <= day <= 23: print("Весы")
        elif 24 <= day <= 31: print("Скорпион")
        else: print("Нет такой даты")

    case "Ноябрь" | "ноябрь" | "11" | "11":
        if 1 <= day <= 22: print("Скорпион")
        elif 23 <= day <= 30: print("Стрелец")
        else: print("Нет такой даты")

    case "Декабрь" | "декабрь" | "12" | "12":
        if 1 <= day <= 21: print("Стрелец")
        elif 22 <= day <= 31: print("Козерог")
        else: print("Нет такой даты")

    case _:
        print("Нет такого месяца")
```

```python
width = int(input("Введите ширину товара, см: "))
length = int(input("Введите длину товара, см: "))
height = int(input("Введите высоту товара, см: "))

if width > 200 or length > 200 or height > 200:
    print("Упаковка для лыж")
elif ((15 < width < 50) or (15 < length < 50) or (15 < height < 50)):
    print("Коробка №2")
elif width <= 15 and length <= 15 and height <= 15:
    print("Коробка №1")
else: print("Коробка №3")
```

```python
number = int(input("Введите номер билета: "))
left = number//1000
right = number%1000

print("Билет счастливый") if (left//100 + left%100//10 + left%10 == right//100 + right%100//10 + right%10) else print("Билет несчастливый")
```

```python
figure = str(input("Введите тип фигуры: "))

match figure:
    case "Круг" | "круг":
        radius = int(input("Введите радиус круга: "))
        print("Площадь круга: ", 3.14*radius**2)

    case "Треугольник" | "треугольник":
        side_A = int(input("Введите длину стороны А: "))
        side_B = int(input("Введите длину стороны B: "))
        side_C = int(input("Введите длину стороны C: "))
        p = (side_A + side_B + side_C)/2                    # полупериметр треугольника
        print("Площадь треугольника: ", (p*(p-side_A)*(p-side_B)*(p-side_C))**0.5)		# по формуле Герона

    case "Прямоугольник" | "прямоугольник":
        side_A = int(input("Введите длину стороны А: "))
        side_B = int(input("Введите длину стороны B: "))
        print("Площадь прямоугольника: ", side_A * side_B)

    case _:
        print("Площадь этой фигуры я не умею вычислять")
```
