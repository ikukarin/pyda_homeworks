```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import time

queries_list = ['python', 'анализ данных']
URL = 'https://habr.com/ru/search/'


def get_search_results(queries):
    df = pd.DataFrame()
    for query in queries:
        params = {
            'q': query
            }

        req = requests.get(URL, params=params)

        time.sleep(0.3)

        soup = BeautifulSoup(req.text)
        articles = soup.find_all('article', class_='tm-articles-list__item')

        for article in articles:
            title = article.find('h2', class_='tm-title tm-title_h2').text.strip()
            link = article.find('h2', class_='tm-title tm-title_h2').find('a').get('href')
            date = article.find('time')['title']
            row = {'date': date, 'title': title, 'link': link}
            df = pd.concat([df, pd.DataFrame([row])])
    return df.reset_index(drop=True).drop_duplicates(subset=['title'])

get_search_results(queries_list)
```
