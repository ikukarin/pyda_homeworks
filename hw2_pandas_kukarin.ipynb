```python
import pandas as pd

ratings = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/ratings.csv')

def film_classificator(line):
    if line['rating'] <= 2:
        return 'Низкий рейтинг'
    elif 2 < line['rating'] <= 4:
        return 'Средний рейтинг'
    else:
        return 'Высокий рейтинг'

ratings['film_class'] = ratings.apply(film_classificator, axis=1)
ratings.head()
```

```python
geo_data = {

'Центр':['москва', 'тула', 'ярославль'],

'Северо-Запад':['петербург', 'псков', 'мурманск'],

'Дальний Восток':['владивосток', 'сахалин', 'хабаровск']

}

df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/keywords.csv')

def set_region(row):
    q = row['keyword'].split()
    for region, city_list in geo_data.items():
        for word in q:
            if word in city_list:
                return region
    return 'undefined'


df['region'] = df.apply(set_region, axis=1)
df[(df['region']!='undefined')].head(10)
```

```python
movies = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/movies.csv')
ratings = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/ratings.csv')

years_list = [str(year) for year in range(1950, 2011)]

def get_production_year(row):
    for year in years_list:
        if year in row['title']:
            return year
    return 1900

movies['year'] = movies.apply(get_production_year, axis=1)
data = ratings.merge(movies, on='movieId', how="left")
data[['year', 'rating']].groupby('year').mean().sort_values('rating', ascending=False)
```
