```python
import json
from datetime import datetime

purchases = {}
with open('purchase_log.txt', 'r', encoding='UTF-8') as f1:
    for line in f1:
        line = json.loads(line.strip())
        key = line['user_id']
        value = line['category']
        if key != 'user_id':
            purchases.update({key: value})

with open('visit_log.csv', 'r', encoding='UTF-8') as f2, open('funnel.csv', 'w', encoding='UTF-8') as f3:
    for line in f2:
        id_user = line.strip().split(',')[0]
        if id_user in purchases.keys():
            result_data = line.strip() + ',' + purchases[id_user] + '\n'
            f3.write(result_data)
```
